# Bregman–KL projection

Minimize KL(p||q) on the affine constraint p1=p2. The Pythagorean equality holds for *every* feasible p, because the projection lands in the relative interior of an affine set.

```python
from iglab.categorical import kl
from iglab.duality import kl_projection_equal_first_two
import numpy as np
q=np.array([.6,.3,.1]);r=kl_projection_equal_first_two(q)
for a in [.1,.25,.4]:
 p=np.array([a,a,1-2*a])
 print(a,kl(p,q),kl(p,r)+kl(r,q))
```
