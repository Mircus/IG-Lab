# Čencov and the metric forced by refinement

**Claim:** split a rational probability vector into equal cells. Fisher norms are preserved; arbitrary noisy channels cannot increase them. This is a computational check, **not** a proof of the theorem.

```python
import numpy as np
from iglab.categorical import *
counts=[2,3,5]
p=np.array(counts)/sum(counts)
u=np.array([.2,-.1,-.1])
K=splitting_channel(counts)
q,v=push_forward(K,p,u)
print("refinement:",fisher_inner(p,u,u),fisher_inner(q,v,v))
print("recovery:",np.allclose(deterministic_recovery(counts)@K,np.eye(3)))
```

```python
noisy=np.array([[.8,.2,.5],[.2,.8,.5]])
print("noisy-channel contraction:", fisher_contraction_gap(noisy,p,u))
```
